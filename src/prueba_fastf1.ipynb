```python

# Cargar la carrera del GP de España 2024
session = fastf1.get_session(2024, "Spanish", "R")
session.load()

# Datos de todas las vueltas
laps = session.laps

print("Número de vueltas registradas:", len(laps))
print()
print("Columnas disponibles:")
print(laps.columns.tolist())
print()
print("Primeras 5 vueltas:")
print(laps.head())

print()
print("Ejemplo de una vuelta:")
print(laps.iloc[100].to_string())
```

```python
print()
print("Vueltas eliminadas:")
print(laps["Deleted"].value_counts())

print()
print("Vueltas precisas:")
print(laps["IsAccurate"].value_counts())

print()
print("Vueltas con PitOutTime:")
print(laps["PitOutTime"].notna().sum())

print()
print("Vueltas con PitInTime:")
print(laps["PitInTime"].notna().sum())
```

```python
print()
print("Ejemplos de vueltas no precisas:")
print(
    laps.loc[
        laps["IsAccurate"] == False,
        ["Driver", "LapNumber", "LapTime", "Compound", "TyreLife", "PitInTime", "PitOutTime"]
    ].head(10)
)
```

```python
clean_laps = laps[
    (laps["Deleted"] == False) &
    (laps["IsAccurate"] == True)
]

print()
print("Vueltas limpias:", len(clean_laps))
```

```python
print()
print("Vueltas con PitOutTime:")
print(clean_laps["PitOutTime"].notna().sum())

print()
print("Vueltas con PitInTime:")
print(clean_laps["PitInTime"].notna().sum())
```

```python
print()
print("Vueltas por piloto:")
print(clean_laps["Driver"].value_counts().sort_index())
```

```python
print()
print("LapTime nulos:", clean_laps["LapTime"].isna().sum())

print()
print("LapTime mínimo:", clean_laps["LapTime"].min())
print("LapTime máximo:", clean_laps["LapTime"].max())
```

```python
clean_laps["LapTimeSeconds"] = clean_laps["LapTime"].dt.total_seconds()

print()
print(clean_laps[["LapTime", "LapTimeSeconds"]].head())
```

```python
categorical_features = [
    "Driver",
    "Team",
    "Compound"
]

numeric_features = [
    "LapNumber",
    "Stint",
    "TyreLife",
    "Position",
    "FreshTyre"
]
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(handle_unknown="ignore")
```

```python
train = clean_laps[clean_laps["LapNumber"] <= 45].copy()
test = clean_laps[clean_laps["LapNumber"] > 45].copy()

print("Train:", len(train))
print("Test:", len(test))

print()
print("Última vuelta de Train:", train["LapNumber"].max())
print("Primera vuelta de Test:", test["LapNumber"].min())
```

```python
X_train = train[features]
y_train = train["LapTimeSeconds"]

X_test = test[features]
y_test = test["LapTimeSeconds"]
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ],
    remainder="passthrough"
)
```

```python
X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Train encoded:", X_train_encoded.shape)
print("Test encoded:", X_test_encoded.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()

print(feature_names)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)
```

```python
model.fit(X_train_encoded, y_train)
```

```python
y_pred = model.predict(X_test_encoded)

print("Primeras predicciones:")
print(y_pred[:10])

print()
print("Valores reales:")
print(y_test.iloc[:10].values)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print(f"MAE: {mae:.3f} segundos")
print(f"RMSE: {rmse:.3f} segundos")
```

```python
baseline_prediction = y_train.mean()

y_pred_baseline = np.full(
    shape=len(y_test),
    fill_value=baseline_prediction
)

mae_baseline = mean_absolute_error(y_test, y_pred_baseline)
rmse_baseline = np.sqrt(mean_squared_error(y_test, y_pred_baseline))

print(f"Baseline MAE: {mae_baseline:.3f} segundos")
print(f"Baseline RMSE: {rmse_baseline:.3f} segundos")
```

```python
import pandas as pd
feature_names = preprocessor.get_feature_names_out()

importances = model.feature_importances_

feature_importance = (
    pd.DataFrame({
        "Feature": feature_names,
        "Importance": importances
    })
    .sort_values("Importance", ascending=False)
)

feature_importance.head(15)
```
