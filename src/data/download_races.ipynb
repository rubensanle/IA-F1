```python
import fastf1
from pathlib import Path


# Carpeta donde guardaremos los datos
OUTPUT_DIR = Path("data/raw")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


# Obtener calendario 2024
schedule = fastf1.get_event_schedule(2024)

# Solo carreras oficiales
races = schedule[
    schedule["RoundNumber"].between(1, 24)
].copy()


for _, race in races.iterrows():

    round_number = int(race["RoundNumber"])
    event_name = race["EventName"]

    print(f"\n{'=' * 60}")
    print(f"Descargando R{round_number}: {event_name}")
    print(f"{'=' * 60}")

    try:
        # Cargar sesión de carrera
        session = fastf1.get_session(
            2024,
            round_number,
            "R"
        )

        session.load()

        # Obtener vueltas
        laps = session.laps

        # Nombre del archivo
        output_file = OUTPUT_DIR / f"2024_R{round_number:02d}.parquet"

        # Guardar
        laps.to_parquet(output_file, index=False)

        print(f"OK: {len(laps)} vueltas")
        print(f"Guardado en: {output_file}")

    except Exception as e:

        print(f"ERROR en R{round_number}: {event_name}")
        print(f"Motivo: {e}")
```
