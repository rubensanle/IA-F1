```python
import fastf1
from pathlib import Path
```

```python
YEAR = 2024

RAW_DIR = Path(f"data/raw/{YEAR}")
RAW_DIR.mkdir(parents=True, exist_ok=True)


# Todas las sesiones que nos interesa guardar.
SESSIONS = [
    "FP1",
    "FP2",
    "FP3",
    "SQ",
    "S",
    "Q",
    "R",
]


# Obtenemos el calendario de 2024
schedule = fastf1.get_event_schedule(YEAR)

# Eliminamos pre-season testing
races = schedule[
    schedule["RoundNumber"].between(1, 24)
].copy()


for _, event in races.iterrows():

    round_number = int(event["RoundNumber"])
    event_name = event["EventName"]

    # Nombre seguro para la carpeta
    folder_name = (
        f"R{round_number:02d}_"
        f"{event_name.replace(' Grand Prix', '').replace(' ', '_')}"
    )

    output_dir = RAW_DIR / folder_name
    output_dir.mkdir(parents=True, exist_ok=True)

    print("\n" + "=" * 70)
    print(f"R{round_number:02d} - {event_name}")
    print("=" * 70)

    for session_code in SESSIONS:

        output_file = output_dir / f"{session_code}.parquet"

        # Si ya existe, no volvemos a descargarlo
        if output_file.exists():
            print(f"[SKIP] {session_code} ya existe")
            continue

        print(f"\nDescargando {session_code}...")

        try:

            session = fastf1.get_session(
                YEAR,
                round_number,
                session_code
            )

            session.load()

            laps = session.laps

            # Algunas sesiones pueden no existir
            # en fines de semana Sprint.
            if laps is None or laps.empty:
                print(f"[SKIP] {session_code} sin datos")
                continue

            laps.to_parquet(
                output_file,
                index=False
            )

            print(
                f"[OK] {session_code}: "
                f"{len(laps)} vueltas"
            )

        except Exception as e:

            print(
                f"[SKIP] {session_code} no disponible"
            )

            print(f"       Motivo: {e}")


print("\n" + "=" * 70)
print("DESCARGA TERMINADA")
print("=" * 70)
```

```python
from pathlib import Path
import pandas as pd


YEAR = 2024
RAW_DIR = Path(f"data/raw/{YEAR}")


print("=" * 70)
print(f"DATASET {YEAR}")
print("=" * 70)


race_dirs = sorted(
    [p for p in RAW_DIR.iterdir() if p.is_dir()]
)


print(f"\nGPs encontrados: {len(race_dirs)}")


total_files = 0
total_laps = 0


for race_dir in race_dirs:

    print("\n" + "-" * 70)
    print(race_dir.name)
    print("-" * 70)

    session_files = sorted(race_dir.glob("*.parquet"))

    for file in session_files:

        try:
            df = pd.read_parquet(file)

            rows = len(df)

            print(
                f"{file.stem:<5} "
                f"{rows:>5} vueltas"
            )

            total_files += 1
            total_laps += rows

        except Exception as e:

            print(
                f"{file.stem:<5} ERROR: {e}"
            )


print("\n" + "=" * 70)
print("RESUMEN")
print("=" * 70)

print(f"GPs:             {len(race_dirs)}")
print(f"Archivos:        {total_files}")
print(f"Filas/vueltas:   {total_laps:,}")
```
