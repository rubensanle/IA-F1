```python
from pathlib import Path

import pandas as pd
```

```python


INPUT_PATH = Path("data/processed/race_laps_2024.parquet")
OUTPUT_PATH = Path("data/processed/stint_laps_2024.parquet")


def main():
    print("Cargando dataset de carreras...")
    df = pd.read_parquet(INPUT_PATH)

    print(f"Filas originales: {len(df):,}")

    # Ordenamos las vueltas dentro de cada stint
    df = df.sort_values(
        [
            "Race",
            "Driver",
            "Stint",
            "LapNumber",
        ]
    ).copy()

    # Número de vuelta dentro del stint
    df["StintLap"] = (
        df.groupby(
            ["Race", "Driver", "Stint"],
            sort=False
        ).cumcount()
        + 1
    )

    # Columnas que necesitamos para estudiar los stints
    columns = [
        "Race",
        "Driver",
        "DriverNumber",
        "Team",
        "Compound",
        "Stint",
        "StintLap",
        "LapNumber",
        "TyreLife",
        "LapTimeSeconds",
        "Position",
        "FreshTyre",
        "TrackStatus",
    ]

    df = df[columns]

    # Guardamos
    OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
    df.to_parquet(OUTPUT_PATH, index=False)

    print()
    print("=" * 60)
    print("DATASET DE STINTS CREADO")
    print("=" * 60)

    print(f"Filas: {len(df):,}")
    print(f"Columnas: {len(df.columns)}")
    print(f"Carreras: {df['Race'].nunique()}")
    print(f"Drivers: {df['Driver'].nunique()}")
    print(f"Stints: {df.groupby(['Race', 'Driver', 'Stint']).ngroups}")

    print()
    print("Columnas:")
    print(df.columns.tolist())

    print()
    print("Primeras filas:")
    print(df.head(10).to_string(index=False))

    print()
    print(f"Guardado en: {OUTPUT_PATH}")


if __name__ == "__main__":
    main()
```
