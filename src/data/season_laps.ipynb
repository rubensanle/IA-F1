```python
from pathlib import Path

import pandas as pd
```

```python
INPUT_DIR = Path("data/raw/2024")
OUTPUT_PATH = Path("data/processed/session_laps_2024.parquet")


def main():
    all_sessions = []

    # Recorremos cada GP
    for race_dir in sorted(INPUT_DIR.iterdir()):
        if not race_dir.is_dir():
            continue

        # Recorremos las sesiones disponibles en ese GP
        for session_path in sorted(race_dir.glob("*.parquet")):
            session = session_path.stem

            print(f"Cargando: {race_dir.name} / {session}")

            df = pd.read_parquet(session_path)

            # Añadimos información que identifica la sesión
            df["Race"] = race_dir.name
            df["Session"] = session

            all_sessions.append(df)

    if not all_sessions:
        raise RuntimeError("No se encontraron archivos parquet.")

    # Unimos todas las sesiones
    df = pd.concat(all_sessions, ignore_index=True)

    # Ordenamos para facilitar posteriores análisis
    sort_columns = [
        "Race",
        "Session",
    ]

    if "Driver" in df.columns:
        sort_columns.append("Driver")

    if "LapNumber" in df.columns:
        sort_columns.append("LapNumber")

    df = df.sort_values(sort_columns).reset_index(drop=True)

    # Guardamos
    OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
    df.to_parquet(OUTPUT_PATH, index=False)

    print()
    print("=" * 60)
    print("DATASET DE SESIONES CREADO")
    print("=" * 60)

    print(f"Filas: {len(df):,}")
    print(f"Columnas: {len(df.columns)}")
    print(f"GPs: {df['Race'].nunique()}")
    print(f"Sesiones: {df['Session'].nunique()}")

    print()
    print("Sesiones:")
    print(df["Session"].value_counts().sort_index())

    print()
    print("Filas por sesión:")
    print(
        df.groupby("Session")
        .size()
        .sort_index()
    )

    print()
    print(f"Guardado en: {OUTPUT_PATH}")


if __name__ == "__main__":
    main()

```
