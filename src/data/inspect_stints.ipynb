```python
import pandas as pd
from pathlib import Path
```

```python

FILE = Path("data/processed/race_laps_2024.parquet")

df = pd.read_parquet(FILE)


print("=" * 70)
print("ANÁLISIS DE STINTS")
print("=" * 70)


# ------------------------------------------------------------
# Número de stints
# ------------------------------------------------------------

print("\nNúmero de stints:")

stints = (
    df.groupby(
        ["Race", "Driver", "Stint"]
    )
    .agg(
        Compound=("Compound", "first"),
        Laps=("LapNumber", "count"),
        FirstLap=("LapNumber", "min"),
        LastLap=("LapNumber", "max"),
        FirstTyreLife=("TyreLife", "min"),
        LastTyreLife=("TyreLife", "max"),
    )
    .reset_index()
)


print(f"Total de stints: {len(stints):,}")


# ------------------------------------------------------------
# Distribución de duración
# ------------------------------------------------------------

print("\nDuración de los stints:")

print(
    stints["Laps"]
    .describe()
)


# ------------------------------------------------------------
# Stints por compuesto
# ------------------------------------------------------------

print("\nStints por compuesto:")

print(
    stints["Compound"]
    .value_counts(dropna=False)
)


# ------------------------------------------------------------
# Duración media por compuesto
# ------------------------------------------------------------

print("\nDuración media por compuesto:")

print(
    stints.groupby("Compound")["Laps"]
    .agg(["count", "mean", "median", "min", "max"])
    .sort_values("mean", ascending=False)
)


# ------------------------------------------------------------
# Ejemplos
# ------------------------------------------------------------

print("\nPrimeros 20 stints:")

print(
    stints.head(20).to_string(index=False)
)


print("\n" + "=" * 70)
```
